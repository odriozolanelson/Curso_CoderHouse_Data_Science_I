## Bloque 4 Funciones personalizadas

La función clasificar_corredor recorre las transacciones y las clasifica según la ubicación de los bancos y las monedas de envío y recepción. Distingue ubicaciones faltantes, operaciones internacionales y operaciones nacionales con monedas iguales o distintas. Esta clasificación es descriptiva y no determina por sí sola que una transacción sea ilícita.
La función lambda clasifica el monto de cada transacción como bajo (menor a 5.000), medio (desde 5.000 y menor a 15.000) o alto (15.000 o más). Los límites son rangos definidos para esta exploración con el objetivo de agrupar las transacciones por monto.
Ambas clasificaciones se agregan al DataFrame como las columnas Tipo_corredor y Categoria_monto.


```python
def clasificar_corredor(origen, destino, moneda_envio, moneda_recepcion):
    
    if pd.isna(origen) or pd.isna(destino):
        return "Ubicación faltante"
    elif origen != destino:
        return "Operación Internacional"
    elif pd.isna(moneda_envio) or pd.isna(moneda_recepcion):
        return "Moneda faltante"
    elif moneda_envio != moneda_recepcion:
        return "Operación Nacional, distinta moneda"
    else:
        return "Operación Nacional, misma moneda"


# Columnas que se usarán, en el orden en que se pasan a la función.
columnas = [
    "Sender_bank_location",
    "Receiver_bank_location",
    "Payment_currency",
    "Received_currency"
]

resultados = []

# Recorre cada fila de esas cuatro columnas.
for fila in df[columnas].itertuples(index=False, name=None):
    resultado = clasificar_corredor(*fila)
    resultados.append(resultado)

# Se agrega la clasificación al DataFrame.
df["Tipo_corredor"] = resultados

# Lambda: asigna una categoría según el monto de la transacción.
clasificar_monto = lambda monto: (
    "Monto bajo" if monto < 5000
    else "Monto medio" if monto < 15000
    else "Monto alto"
)

df["Categoria_monto"] = df["Amount"].apply(clasificar_monto)

# Se muestran 10 filas en forma aleatoria de las columnas que nos interesan y la columna creada,para comprobar el resultado.
display(df[columnas + ["Tipo_corredor"] + ["Categoria_monto"]].sample(10))
```

6. Porcentaje de transacciones según tipo de pago y ubicación bancaria

```python
# Creamos df_graficos con las columnas necesarias.
df_graficos = df_sin_tipologia[
    [
        "Sender_bank_location",
        "Receiver_bank_location",
        "Payment_type"
    ]
].copy()

# Creamos una categoría para comparar las ubicaciones bancarias.
origen = df_graficos["Sender_bank_location"]
destino = df_graficos["Receiver_bank_location"]

df_graficos["Comparacion_ubicacion"] = np.select(
    [
        origen.isna() | destino.isna(),
        origen.eq(destino)
    ],
    [
        "Ubicación faltante",
        "Misma ubicación"
    ],
    default="Ubicaciones distintas"
)

df_graficos["Payment_type_grafico"] = (
    df_graficos["Payment_type"].fillna("Sin dato")
)

# Calculamos los porcentajes dentro de cada tipo de pago.
tabla_ubicacion = pd.crosstab(
    df_graficos["Payment_type_grafico"],
    df_graficos["Comparacion_ubicacion"],
    normalize="index"
) * 100

tabla_ubicacion = (
    tabla_ubicacion
    .reset_index()
    .melt(
        id_vars="Payment_type_grafico",
        var_name="Comparación de ubicaciones",
        value_name="Porcentaje"
    )
)

fig, ax = plt.subplots(figsize=(11, 6))

sns.barplot(
    data=tabla_ubicacion,
    x="Payment_type_grafico",
    y="Porcentaje",
    hue="Comparación de ubicaciones",
    ax=ax
)

ax.set_title("Ubicaciones bancarias según el tipo de pago")
ax.set_xlabel("Tipo de pago")
ax.set_ylabel("Porcentaje de transacciones")
plt.xticks(rotation=30, ha="right")

# Movemos la leyenda debajo del gráfico.
elementos, etiquetas = ax.get_legend_handles_labels()
ax.legend_.remove()

fig.legend(
    elementos,
    etiquetas,
    title="Comparación de ubicaciones",
    loc="lower center",
    bbox_to_anchor=(0.5, 0.01),
    ncol=len(etiquetas)
)

fig.tight_layout(rect=[0, 0.15, 1, 0.95])

fig.savefig(
    "ubicaciones_por_tipo_de_pago.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

En casi todos los tipos de pago predominan las transacciones con la misma ubicación bancaria. En Cross-border, en cambio, todas aparecen con ubicaciones distintas, algo coherente con el significado de esa categoría. Esta comparación describe el dataset y no indica por sí sola que una operación sea sospechosa.

7. Relación entre el monto, la ubicación bancaria y la moneda de las transacciones

```python
# Creamos un DataFrame separado con las columnas necesarias.
df_correlacion = df_sin_tipologia[
    [
        "Amount",
        "Sender_bank_location",
        "Receiver_bank_location",
        "Payment_currency",
        "Received_currency"
    ]
].copy()

df_correlacion["Amount"] = pd.to_numeric(
    df_correlacion["Amount"],
    errors="coerce"
)

# Creamos los indicadores auxiliares.
df_correlacion["Ubicacion_distinta"] = (
    df_correlacion["Sender_bank_location"].fillna("")
    != df_correlacion["Receiver_bank_location"].fillna("")
).astype(int)

df_correlacion["Moneda_distinta"] = (
    df_correlacion["Payment_currency"].fillna("")
    != df_correlacion["Received_currency"].fillna("")
).astype(int)

columnas_correlacion = [
    "Amount",
    "Ubicacion_distinta",
    "Moneda_distinta"
]

# Calculamos la correlación usando todas las filas disponibles.
matriz_correlacion = df_correlacion[columnas_correlacion].corr()

matriz_correlacion = matriz_correlacion.rename(
    index={
        "Amount": "Monto",
        "Ubicacion_distinta": "Ubicación distinta",
        "Moneda_distinta": "Moneda distinta"
    },
    columns={
        "Amount": "Monto",
        "Ubicacion_distinta": "Ubicación distinta",
        "Moneda_distinta": "Moneda distinta"
    }
)

fig, ax = plt.subplots(figsize=(8, 6))

sns.heatmap(
    matriz_correlacion,
    annot=True,
    cmap="vlag",
    center=0,
    vmin=-1,
    vmax=1,
    fmt=".2f",
    ax=ax
)

ax.set_title("Relación entre el monto, la ubicación y la moneda")

plt.tight_layout()
fig.savefig(
    "heatmap_correlacion_dataset_completo.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

El mapa muestra una asociación positiva alta entre las transacciones con ubicación bancaria distinta y las que tienen monedas distintas (0,91): ambas características suelen presentarse juntas en el conjunto de datos. En cambio, el monto tiene una correlación lineal muy baja con esos indicadores (0,02). La correlación no implica causalidad; los valores de 1,00 en la diagonal representan cada variable correlacionada consigo misma.
